### - What is a Spark Window function?
windows function performs calculation across a group of rows(windows) while  keeping every row in the output. unlike groupby() it does not reduce the number of rows. 

### - groupBy() vs Window — what's the difference?
    **Windows Function** -->  perform calculations across a group of rows while **keeping every row** in the output.

    **_groupBy()_** --> Groups rows by one or more keys and, when followed by aggregation, produces aggregated results with fewer rows.


### - Difference between row_number(), rank() and dense_rank()?

    1. **row_number()**
        - assigns a unique sequential number to each row within a partition according to the specified ordering. It can be used for latest-record selection and deduplication.
        - getlatest record per key
        - Similar to SQL row_number()

    2. **rank()**
        if 2 rows has same rnk , it skips the next number (1,2,2,4)

    3. **dense_rank()**
         similar to rank() but does not skip the next number. (1,2,2,3)
         

### - How would you retrieve the latest record for every customer?
Partition the data by custId, order each partition by the timestamp descending, apply row_number(), and filter for row_number = 1.

### - How would you deduplicate records when multiple versions exist?
- For **_exact duplicate rows_**, use distinct() or dropDuplicates()
- If there are **_multiple versions_** of the same customer, partition by the business key, order by the timestamp descending, apply _row_number_(), and retain row 1.

### - What does lag() do?
lag() retrieves a value from a preceding row within the defined window without removing the current row. It's useful for comparing current and previous values.
dfPrevStatus = dfCustomerCDC1.withColumn("prev_status", lag(col("status"),1).over(custPartition)) 


### - Give a real ETL use case for lag().
In a customer billing pipeline, we can use lag() to compare the current billing cycle's sales amount against the previous billing cycle. We can calculate the percentage change and flag cases where the difference exceeds a business threshold such as ±4%.

### - Can window operations cause a shuffle? Why?
Yes. Window operations can cause a shuffle because Spark may need to redistribute records across partitions so rows with the same partition key are processed together. Spark then sorts records within each partition according to the window's orderBy. Shuffles can be expensive because they involve data movement across executors/network.
````
eg-
join()
    ↓
may cause SHUFFLE

groupBy()
    ↓
may cause SHUFFLE

Window.partitionBy()
    ↓
may cause SHUFFLE

distinct()
    ↓
may cause SHUFFLE

repartition()
    ↓
causes SHUFFLE

Notice the pattern?
````

Whenever Spark needs to **reorganize data** so records with the same key can be processed together, think about shuffle.
