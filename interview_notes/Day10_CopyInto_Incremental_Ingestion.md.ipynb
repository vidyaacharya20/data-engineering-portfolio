# Incremental File ingestion using COPY INTO

**1. What is COPY INTO?**
COPY INTO is a **Databricks SQL command **that loads files from a source location into a Delta table.
Its main advantage is **_idempotent file loading:_** by default, it tracks **files previously loaded into the target table** and skips them on subsequent executions.

````
COPY INTO target_table
FROM 'source_path'
FILEFORMAT = PARQUET;

````

**Idempotency**
An operation is idempotent when rerunning it produces the same intended result rather than accumulating unintended changes.

For example, rerunning COPY INTO against unchanged source files should not append the same files again.

**Q1. What's the difference between Auto Loader and COPY INTO?**
Both support incremental file ingestion in Databricks. 
- Auto Loader --> uses Structured Streaming and checkpoints to track processing progress, while 
- COPY INTO --> is a SQL command that tracks previously loaded files through the Delta table. 
I would generally choose COPY INTO for simpler scheduled batch ingestion and Auto Loader for more scalable, frequently arriving data.

**Q2. Is COPY INTO idempotent?**
Yes, by default COPY INTO tracks previously loaded files and skips them when the same command is executed again. 
However, this is file-level idempotency. 
If new files contain duplicate business records, those duplicates can still be loaded.

**Q3. Does COPY INTO require a checkpoint?**
No. Unlike Auto Loader, COPY INTO doesn't require a separate streaming checkpoint location. It maintains file-loading information through the Delta table.

**When would you choose COPY INTO instead of Auto Loader?**
I would choose COPY INTO when I have a relatively straightforward batch ingestion requirement, such as loading Parquet or CSV files from cloud storage into a Delta table on a schedule. It tracks previously loaded files automatically, so I don't need to maintain a streaming checkpoint. For larger-scale or continuously arriving files, I'd generally prefer Auto Loader because it provides scalable file discovery, streaming integration, and more flexible schema evolution handling.
