**1. Why are Spark DataFrames immutable?**
PySpark DataFrames are immutable - operations like withColumn(), drop(), filter() don't change the original DataFrame. _**They return a new one.**_ You must reassign if you want to keep the changes.

Example:
withColumn() creates a new DataFrame with the added column - it doesn't modify dfCustomers itself

You display the new DataFrame (with country), but don't save it to a variable

Then you write the original dfCustomers (without country) to the table

  ```
  python
  # Import lit function to create literal/constant values
  from pyspark.sql.functions import lit

  # Add a new column 'country' with constant value 'USA' to dfCustomers
  # withColumn() creates a NEW DataFrame - must reassign to keep the change
  dfCustomers = dfCustomers.withColumn("country", lit("USA"))

  # Save the updated DataFrame to the customers table
  # mode("overwrite") - replaces existing table data
  # option("mergeSchema", "true") - allows adding new columns to existing table schema
  dfCustomers.write.mode("overwrite").option("mergeSchema", "true").saveAsTable("customers")

  # Display the updated DataFrame with the new country column
  disp
  ````

**4. How would you handle NULL values in PySpark?**

  First I determine what the NULL means from a business perspective. For mandatory fields, I may reject or filter the record. For optional fields, I may keep the NULL or replace it with an appropriate default. PySpark provides functions like isNull(), isNotNull(), fill(), drop(), and when() to handle these situations.

- Remove Rows containing Null using **drop() **--> df.na.drop()
- Fill Nulls
    - Use **conditional check** to fill NULL
       
        ````
        python
        df.withColumn(
            "state",
            when(col("state").isNull(), "UNKNOWN")
            .otherwise(col("state"))
           )
           ````

    - Filter Them
        ````
        python
        df.**filter**(col("state").isNotNull())
        ````

    - **fillna()** --> subsitute null with a value
    - **coaleses()** --> if null the exp2 become the value
